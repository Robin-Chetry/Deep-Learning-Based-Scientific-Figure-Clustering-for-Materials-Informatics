import all necessary libraries

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
```

load features

```python
# =========================================================
# LOAD FEATURES
# =========================================================

features = np.load(
    "../embeddings/efficientnet_b0_features.npy"
)
print("Original feature shape:")
print(features.shape)
```

apply pca

```python
# =========================================================
# APPLY PCA
# =========================================================

# keep 95% variance
# pca = PCA(n_components=0.95)
pca=PCA(n_components=200)

pca_features = pca.fit_transform(features)
```

results

```python
# =========================================================
# RESULTS
# =========================================================

print("\nPCA feature shape:")
print(pca_features.shape)

print("\nNumber of PCA components:")
print(pca.n_components_)
```

explained variance and plotting it 

```python
# =========================================================
# EXPLAINED VARIANCE
# =========================================================

explained_variance = (
    pca.explained_variance_ratio_
)

# cumulative variance
cumulative_variance = np.cumsum(
    explained_variance
)

# =========================================================
# PLOT VARIANCE
# =========================================================

plt.figure(figsize=(10,5))

plt.plot(cumulative_variance)

plt.xlabel("Number of Components")

plt.ylabel("Cumulative Explained Variance")

plt.title("PCA Explained Variance")

plt.grid(True)

plt.show()
```

saving the pca features

```python
# =========================================================
# SAVE PCA FEATURES
# =========================================================

np.save(
    "../embeddings/pca_features.npy",
    pca_features
)

print("\nPCA features saved!")
```
