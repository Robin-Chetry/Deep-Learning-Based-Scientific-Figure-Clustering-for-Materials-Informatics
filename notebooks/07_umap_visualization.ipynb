load the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import umap
```

load the pca features

```python
# =========================================================
# LOAD PCA FEATURES
# =========================================================

pca_features = np.load(
    "../embeddings/pca_features.npy"
)

print("PCA feature shape:")
print(pca_features.shape)
```

load umap model

```python
# =========================================================
# CREATE UMAP MODEL
# =========================================================

umap_model = umap.UMAP(

    n_components=2,

    n_neighbors=200,

    min_dist=0.001,

    metric='cosine',

    random_state=3
)
```

```python
# =========================================================
# FIT + TRANSFORM
# =========================================================

umap_embeddings = umap_model.fit_transform(
    pca_features
)

print("\nUMAP embedding shape:")
print(umap_embeddings.shape)
```

```python
# =========================================================
# PLOT UMAP
# =========================================================

plt.figure(figsize=(12,10))

plt.scatter(

    umap_embeddings[:,0],

    umap_embeddings[:,1],

    s=5,

    alpha=0.7
)

plt.title(
    "UMAP Visualization of Scientific Images"
)

plt.xlabel("UMAP Dimension 1")

plt.ylabel("UMAP Dimension 2")

plt.show()
```

saving the umap embeddings

```python
# =========================================================
# SAVE UMAP EMBEDDINGS
# =========================================================

np.save(

    "../embeddings/umap_embeddings.npy",

    umap_embeddings
)

print("\nUMAP embeddings saved!")
```
